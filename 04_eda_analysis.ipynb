```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/fraudTrain_featured.csv")
```

```python
df["trans_date_trans_time"] = pd.to_datetime(df["trans_date_trans_time"])
```

```python
df.shape
```

```python
total_transaction = len(df)
fraud_transaction = df["is_fraud"].sum()
fraud_rate = (
    fraud_transaction / total_transaction
)*100
print("Total Transaction:",total_transaction)
print("Fraud Transaction:",fraud_transaction)
print("Fraud Rate:",round(fraud_rate,2),"%")
```

```python
fraud_by_hour = df.groupby("transaction_hour").agg(total_transaction=("is_fraud","count"),
                                                   fraud_transaction = ("is_fraud","sum"))
```

```python
fraud_by_hour["fraud_rate"]= (fraud_by_hour["fraud_transaction"]/fraud_by_hour["total_transaction"])*100
```

```python
fraud_by_hour.sort_values("fraud_rate",ascending = False).head(10)
```

```python
fraud_by_day = df.groupby("day_of_week").agg(total_transaction=("is_fraud","count"),
                                             fraud_transaction=("is_fraud","sum"))
```

```python
fraud_by_day["fraud_rate"]=(fraud_by_day["fraud_transaction"]/fraud_by_day["total_transaction"])*100
```

```python
fraud_by_day.sort_values("fraud_rate",ascending = False)
```

```python
fraud_by_amount = df.groupby("amount_band").agg(total_transaction=("is_fraud","count"),
                                                fraud_transaction=("is_fraud","sum"))
```

```python
fraud_by_amount["fraud_rate"]= (fraud_by_amount["fraud_transaction"]/fraud_by_amount["total_transaction"])*100
```

```python
fraud_by_amount.sort_values("fraud_rate",ascending = False)
```

```python
fraud_by_state = df.groupby("state").agg(total_transaction =("is_fraud","count"),
                                         fraud_transaction = ("is_fraud","sum"))
```

```python
fraud_by_state["fraud_rate"]= (fraud_by_state["fraud_transaction"]/fraud_by_state["total_transaction"])*100
```

```python
fraud_by_state.sort_values("fraud_rate",ascending = False)
```

```python
state_analysis = fraud_by_state[fraud_by_state["total_transaction"]>=1000].sort_values("fraud_rate",ascending = False)
```

```python
fraud_by_category = (
    df.groupby("category")
      .agg(
          total_transactions=("is_fraud", "count"),
          fraud_transactions=("is_fraud", "sum")
      )
)

```

```python
fraud_by_category["fraud_rate"] = (
    fraud_by_category["fraud_transactions"]
    / fraud_by_category["total_transactions"]
) * 100

```

```python
fraud_by_category.sort_values(
    "fraud_rate",
    ascending=False
)
```

```python
fraud_hour_category = df.groupby(["transaction_hour","category"]).agg(total_transactions=("is_fraud","count"),
                                                                    fraud_transactions=("is_fraud","sum"))
```

```python
fraud_hour_category["fraud_rate"] = (fraud_hour_category["fraud_transactions"]/fraud_hour_category["total_transactions"])*100
```

```python
fraud_hour_category.sort_values("fraud_rate",ascending = False).head(20)
```

```python
fraud_hour_category_reliable = fraud_hour_category[
    fraud_hour_category["total_transactions"] >= 500
].sort_values(
    "fraud_rate",
    ascending=False
)

fraud_hour_category_reliable.head(20)
```
