```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/fraudTrain_cleaned.csv")

df["trans_date_trans_time"] = pd.to_datetime(
    df["trans_date_trans_time"]
)

df["dob"] = pd.to_datetime(df["dob"])

print(df.shape)
```

```python
df["transaction_hour"] = df["trans_date_trans_time"].dt.hour
```

```python
df["transaction_hour"].value_counts().sort_index()
```

```python
df["day_of_week"]=df["trans_date_trans_time"].dt.day_name()
```

```python
df["day_of_week"].value_counts()
```

```python
df["month"]=df["trans_date_trans_time"].dt.month
```

```python
df["month_name"]=df["trans_date_trans_time"].dt.month_name()
```

```python
df["customer_age"]=((df["trans_date_trans_time"]-df["dob"]).dt.days/365.25).astype(int)
```

```python
df["customer_age"].describe()
```

```python
df["is_weekend"] = df["day_of_week"].isin(["Saturday","Sunday"])
```

```python
df["is_weekend"].value_counts()
```

```python
bins = [0,50,100,500,1000,5000,10000,float("inf")]
labels = ["0-50","50-100","100-500","500-1k","1k-5k","5k-10k","10k"]
df["amount_band"] = pd.cut(df["amt"],bins=bins,labels=labels,include_lowest = True)
```

```python
df["amount_band"].value_counts()
```

```python
df.to_csv("../data/fraudTrain_featured.csv", index=False)
```
